```python
df.drop(columns = ["Unnamed: 12", 'Unnamed: 13', 'Unnamed: 14', 'Unnamed: 15', 'Unnamed: 16', 'Unnamed: 17', 'Unnamed: 18', 'Unnamed: 19', 'Unnamed: 20', 'Unnamed: 21', 'Unnamed: 22', 'Unnamed: 23', 'Unnamed: 24', 'Unnamed: 25'], inplace = True)
```

```python
df['director'].fillna('Unknown', inplace = True)
```

```python
df['cast'].fillna('Unknown', inplace = True)
```

```python
df['country'].fillna('Unknown', inplace = True)
```

```python
df.info()
```

```python
df.dropna(subset = ['date_added', 'rating', 'duration'], inplace = True)
```

```python
df.info()
```

```python
df['type'].unique()
```

```python
x = df['type'].value_counts()
```

```python
x
```

```python
def show_count(pct):
    total = sum(x)
    value = int(round(pct*total/100))
    return value
    
l = "Movies", "TV Shows"
plt.pie(x.values, labels = l, autopct = show_count, shadow = bool)
plt.title("Movies Vs Tv Shows", fontweight = "bold")
plt.show()
```

```python
movies = df[(df['type']) == 'Movie']
movies = movies[movies.country != 'Unknown']

top_country_movies = movies.country.value_counts().nlargest(5).sort_values(ascending = False)
```

```python
top_country_movies
```

```python
tvshows = df[(df['type']) == 'TV Show']
tvshows = tvshows[tvshows.country != 'Unknown']

top_country_tvshows = tvshows.country.value_counts().nlargest(5).sort_values(ascending = False)
```

```python
top_country_tvshows
```

```python
products = df[df.country != 'Unknown']
top_movie_show = products.country.value_counts().nlargest(5).sort_values(ascending = False)
top_movie_show
```

```python
director = movies.director.value_counts().nlargest(6).sort_values(ascending = False)
director = director.drop('Unknown')
director
```

```python
tvshow_director = tvshows.director.value_counts().nlargest(6).sort_values(ascending = False)
tvshow_director = tvshow_director.drop('Unknown')
tvshow_director
```
