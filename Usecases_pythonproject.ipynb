```python
#Usecase-2 Display only those customers from CSV_s1, who purchased more than 3 products
```

```python
cust_txn_count = df3.groupby("custno").size().reset_index(name="purchase_count")
result = cust_txn_count[cust_txn_count["purchase_count"] > 3]
result
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=result,
    x="purchase_count",
    y="custno")

plt.title("Customers with More Than 3 Purchases")
plt.xlabel("Number of Products Purchased")
plt.ylabel("Customers")
plt.show()
```

```python
#Usecase-3 Display top 5 most demanded products from CSV_s1
```

```python
top_products = df3.groupby("productno").size().reset_index(name="demand_count")
top_products = top_products.merge(df2, on="productno")
top_products = top_products.sort_values("demand_count", ascending=False).head(5)
print(top_products)
```

```python
#Usecase-7 Display custno, gender, age, profession, contactno, productno, productName,
#txndate, spendby, amount from CSV_s1 for custno = 923
```

```python
#specific_cust = df1[df1['custno'] == 923]['custno','gender','age','profession','contactno','productno','productName','txndate','spendby','amount']
res = df1.merge(df3, on='custno',how ='outer').merge(df2, on='productno',how= 'outer')
```

```python
res.info()
```

```python
# Fix: Wrap the column names in a list using double brackets
cols_to_fix = ['custno', 'age', 'contactNo', 'txnno', 'amount', 'productno']
res[cols_to_fix] = res[cols_to_fix].astype('Int64')
```

```python
filtered_df = res[res['custno'] == 923]
filtered_df
```

```python
filtered_result =filtered_df[['custno', 'gender', 'age', 'profession', 'contactNo', 'productno', 'productName', 
'txndate', 'spendby', 'amount']]
filtered_result
```

```python
# Create the pie chart without labels
ax = filtered_result['productName'].value_counts().plot.pie(
    autopct='%1.1f%%', 
    labels=None,       # Remove labels from the pie
    pctdistance=0.8    # Move percentages slightly out
)

# Move labels to a legend on the right
plt.legend(
    labels=filtered_result['productName'].value_counts().index, 
    loc="center left", 
    bbox_to_anchor=(1, 0, 0.5, 1)
)

plt.title("Unique Products for custno = 923")
plt.ylabel("") # Remove the vertical 'productName' text
plt.show()
```
