```python
top_10 = df1['country'].value_counts().head(10)
```

```python
top_10
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_10.values, y=top_10.index, palette='viridis')
plt.title("Top 10 Countries Producing Content")
plt.xlabel("Count")
plt.ylabel("Country")
plt.show()
```

```python
plt.figure(figsize=(10,6))
top_countries = df1['country'].value_counts().head(10)
sns.barplot(x= df1['type'].values, y=df1['type'].index)
plt.title("Top category")
plt.xlabel("Count")
plt.ylabel("Country")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(x=df1['year_added'], hue=df1['type'], palette='Set1')
plt.xticks(rotation=45)
plt.title("Movies vs TV Shows Added Over Time")
plt.xlabel("Year")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(10,6))
top_10year = df1['year_added'].value_counts().head(10)
sns.barplot(x=top_10year.index, y=top_10year.values, palette='viridis')
plt.title("Top 10 years added Content")
plt.xlabel("Year")
plt.ylabel("Count")
plt.show()
```
