```python
plt.figure(figsize=(8, 5))
sns.histplot(df1['trend_score'], kde=True, color='green', bins=25)

plt.title('Distribution of trend score', fontsize=14)
plt.xlabel(' trend score ')
plt.show()
```

```python

# Create the figure
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# 1. Top Left: Scatter/Regression
# Mapping total_bill -> engagement_score, tip -> trend_score
sns.regplot(ax=axes[0,0], data=df1, x='engagement_score', y='trend_score', scatter_kws={'alpha':0.5})
axes[0,0].set_title('Regression: Engagement Score vs Trend Score')

# 2. Top Right: Pie Chart
# Mapping day -> topic_category
df1['topic_category'].value_counts().plot.pie(ax=axes[0,1], autopct='%1.1f%%', colors=sns.color_palette('pastel'))
axes[0,1].set_ylabel('')
axes[0,1].set_title('Order Distribution by Topic Category')

# 3. Bottom Left: Violin Plot
# Mapping x=time -> source, y=total_bill -> engagement_score, hue=sex -> sentiment
# Check levels for split=True (requires exactly two levels in hue)
if len(df1['sentiment'].unique()) == 2:
    sns.violinplot(ax=axes[1,0], data=df1, x='source', y='engagement_score', hue='sentiment', split=True)
else:
    sns.violinplot(ax=axes[1,0], data=df1, x='source', y='engagement_score', hue='sentiment')
axes[1,0].set_title('Engagement Score Density by Source')
axes[1,0].tick_params(axis='x', rotation=45) # Rotate labels if needed

# 4. Bottom Right: Heatmap (Correlation)
corr = df1.select_dtypes(include=[np.number]).corr()
sns.heatmap(ax=axes[1,1], data=corr, annot=True, cmap='coolwarm')
axes[1,1].set_title('Correlation Matrix')

plt.tight_layout()
plt.savefig('dashboard_analysis.png')
```
