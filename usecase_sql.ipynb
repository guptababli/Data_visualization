```python
import pandas as pd
import sqlite3

# 1. Load your CSV files again to be safe
df_customers = pd.read_csv(r"C:\Users\Babli Gupta\Downloads\PythonProject\CSV_s1\customers.csv")
df_customers.columns = df_customers.columns.str.strip().str.replace("'", "").str.replace('"', "")

df_products= pd.read_csv(r"C:\Users\Babli Gupta\Downloads\PythonProject\CSV_s1\products.csv")
df_products.columns = df_products.columns.str.strip().str.replace("'", "").str.replace('"', "")

df_transactions = pd.read_csv(r"C:\Users\Babli Gupta\Downloads\PythonProject\CSV_s1\transactions.csv")
df_transactions.columns = df_transactions.columns.str.strip().str.replace("'", "").str.replace('"', "")

# 2. Re-establish the connection and register tables
conn = sqlite3.connect(':memory:')
df_customers.to_sql('customers', conn, index=False, if_exists='replace')
df_products.to_sql('products', conn, index=False, if_exists='replace')
df_transactions.to_sql('transactions', conn, index=False, if_exists='replace')


```

```python

# 3. Define your "Report Suite"
queries = {
    "Query 1: top 5": """
        select  * 
        from customers LIMIT 5 ;

    """,
     "Query 2: top 5": """
        select  * 
        from products LIMIT 5 ;

    """,
    "Query 3: top 5": """
        select  * 
        from transactions LIMIT 5 ;

    """,
    "Query 4: DEMAND PRODUCT": """
        SELECT custno, COUNT(productno) AS total_products
        FROM transactions 
        GROUP BY custno 
        HAVING COUNT(productno) > 0;
    """,
     "Query 5:  Display top 5 transactions amount ": """
       SELECT *FROM transactions 
       ORDER BY amount DESC LIMIT 5;
    """,
    "Query 6:Display distinct professions": """
      SELECT DISTINCT profession 
      FROM customers;

    """,
     "Query 7: Display highest age ": """
      SELECT MAX(age)
      FROM customers;
    """,
    
     "Query 8: Display details for custno = 923": """
     SELECT t3.custno ,t3.gender,t3.age,t3.profession,t3.contactNo,
     t2.productName, 
     t1.txnno,t1.amount
     FROM  transactions AS t1
     JOIN products AS t2 ON t1.productno = t2.productno
     JOIN customers AS t3 ON t1.custno = t3.custno
     WHERE t3.custno=923;
    """,
    
     "Query 9: Display details for custno = 923_another way": """
     SELECT 
     c.custno,c.gender,c.age,c.profession,c.contactNo,
     p.productName,
     t.txnno,t.amount
     FROM customers c
     INNER JOIN transactions t
     ON c.custno = t.custno
     INNER JOIN products p
     ON t.productno = p.productno
     WHERE c.custno=923;
    """
    
}
```

```python
# 4. Run the loop
for title, sql in queries.items():
    print(f"--- {title} ---")
    try:
        display(pd.read_sql(sql, conn))
    except Exception as e:
        print(f"Error executing {title}: {e}")
    print("\n" + "="*30 + "\n")
```
